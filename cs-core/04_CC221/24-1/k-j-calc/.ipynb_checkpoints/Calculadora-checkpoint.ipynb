```python
SMOD = [13,16,8,8] # VALORES DE MODOS DEL TIMER EN 0,1 2 Y 3
frecNatural = 11.0592*1e6 #MEGA Hz

while (True):
    F = float(input("Frecuencia del oscilador: "))
    modo = int(input("Modo del timer: "))
    cte = frecNatural/(12*2*F)

    k = cte/2**SMOD[modo]
    j = 2**SMOD[modo] - cte
    
    print("Valor de K = ",round(k)," HEX:",hex(round(k))[2:].upper() + 'h')
    print("Valor de J = ",round(j)," HEX:",hex(round(j))[2:].upper() + 'h')
```
